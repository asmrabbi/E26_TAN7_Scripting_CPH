## Example 3.34.1 — Average satisfaction by service type

**Website locator:** Example 3.34.1

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

valid_satisfaction = df[
    df["satisfaction_score"].between(1, 5)
].copy()

mean_satisfaction_by_service = (
    valid_satisfaction
    .groupby("service_type")["satisfaction_score"]
    .mean()
    .sort_values(ascending=False)
)

print(mean_satisfaction_by_service)
```

## Example 3.34.3 — Several statistics at once with `.agg()`

**Website locator:** Example 3.34.3

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

service_summary = (
    df.groupby("service_type")
    .agg(
        records=("record_id", "count"),
        total_cases=("cases_received", "sum"),
        avg_resolution_days=("resolution_days", "mean"),
        median_resolution_days=("resolution_days", "median"),
        total_complaints=("complaints_received", "sum")
    )
    .reset_index()
)

service_summary
```

## Example 3.38.2 — Recreate tokens if needed

**Website locator:** Example 3.38.2

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

stop_words = set(stopwords.words("english"))
stop_words.difference_update({"not", "no", "nor"})


def prepare_feedback(text):
    if pd.isna(text):
        return []

    tokens = word_tokenize(str(text).lower())

    return [
        token
        for token in tokens
        if token.isalpha()
        and token not in stop_words
    ]


df["feedback_tokens"] = df["feedback"].apply(prepare_feedback)
```

## Example 3.38.3 — Flatten the lists

**Website locator:** Example 3.38.3

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

all_tokens = [
    token
    for token_list in df["feedback_tokens"]
    for token in token_list
]
```

## Example 3.38.4 — Flatten the lists (continued)

**Website locator:** Example 3.38.4

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

all_tokens = []

for token_list in df["feedback_tokens"]:
    for token in token_list:
        all_tokens.append(token)
```

## Example 3.40.1 — Create a helper function

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

def get_tokens_for_subset(dataframe):
    return [
        token
        for token_list in dataframe["feedback_tokens"]
        for token in token_list
    ]
```

## Example 3.40.2 — Create two subsets

## Example 3.40.3 — Create frequency distributions

**Website locator:** Example 3.40.3

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.
import nltk  # Imports the Natural Language Toolkit used for introductory text analysis.
from nltk import FreqDist  # Imports NLTK's frequency-distribution helper.

housing_tokens = get_tokens_for_subset(housing_df)
transport_tokens = get_tokens_for_subset(transport_df)

housing_fdist = FreqDist(housing_tokens)
transport_fdist = FreqDist(transport_tokens)
```

## Example 3.40.4 — Compare raw counts

**Website locator:** Example 3.40.4

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

print("Housing:")
print(housing_fdist.most_common(10))

print("\nTransport:")
print(transport_fdist.most_common(10))
```

## Example 3.40.5 — Compare per 1,000 tokens

**Website locator:** Example 3.40.5

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

def frequency_per_1000(fdist, word):
    if fdist.N() == 0:
        return 0

    return fdist.freq(word) * 1000
```

## Example 3.40.6 — Compare per 1,000 tokens (continued)

**Website locator:** Example 3.40.6

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

word = "helpful"

print(
    "Housing:",
    round(frequency_per_1000(housing_fdist, word), 2)
)

print(
    "Transport:",
    round(frequency_per_1000(transport_fdist, word), 2)
)
```

## Example 3.40.7 — Build a comparison table

**Website locator:** Example 3.40.7

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

comparison_words = [
    "helpful",
    "wait",
    "slow",
    "easy",
    "confusing"
]

rows = []

for word in comparison_words:
    rows.append({
        "word": word,
        "housing_per_1000": frequency_per_1000(
            housing_fdist,
            word
        ),
        "transport_per_1000": frequency_per_1000(
            transport_fdist,
            word
        )
    })

word_comparison = pd.DataFrame(rows)
word_comparison
```

## Example 3.42.11 — Line chart: monthly complaint rate

**Website locator:** Example 3.42.11

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

monthly_summary = (
    df.groupby("report_month")
    .agg(
        total_cases=("cases_received", "sum"),
        total_complaints=("complaints_received", "sum")
    )
    .reset_index()
    .sort_values("report_month")
)

monthly_summary["complaint_rate_pct"] = (
    monthly_summary["total_complaints"]
    / monthly_summary["total_cases"]
    * 100
)
```
