## Safe, repeatable use — the course files stay unchanged

This notebook reads course CSV files from GitHub through **read-only web addresses**. Running a cell downloads a fresh snapshot into the temporary Colab session; it cannot edit or overwrite the file in the course repository.

- Transformations are performed on a separate in-memory working copy.
- The notebook does not mount Google Drive, call the GitHub API, commit, or push.
- Any exported files are student output copies stored in a separate `tutorial_outputs` folder in the current runtime.
- Re-running the notebook starts from the same repository source. In Colab, **Runtime → Restart session and run all** gives a completely fresh demonstration.

You can therefore teach from these notebooks repeatedly without changing the canonical course data.
