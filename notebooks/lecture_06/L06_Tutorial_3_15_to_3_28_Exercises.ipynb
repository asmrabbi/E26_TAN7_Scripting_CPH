````python
# Run this complete block on its own; no earlier cell is required.
import pandas as pd  # Work with tables and missing values.
# Read the canonical synthetic data. Every run starts from this source.
RAW_URL = "https://raw.githubusercontent.com/asmrabbi/E26_TAN7_Scripting_CPH/main/data/E26_TAN7_service_experience_raw.csv"
df_raw = pd.read_csv(RAW_URL)
df = df_raw.copy(deep=True)  # Preserve an unchanged source table.
transformation_log = []

# This teaching file contains a confirmed exact copy of one observation.
rows_before = len(df)
df = df.drop_duplicates().copy()
rows_removed = rows_before - len(df)
transformation_log.append({
    "step": len(transformation_log) + 1, "variable": "all original columns",
    "action": "Remove confirmed exact duplicate", "rationale": "Avoid double-counting",
    "rows_affected": int(rows_removed), "reversible": "Yes, raw source retained",
    "notes": "Repeated IDs with different values remain for review"
})

# Change formatting only. Do not guess whether different labels are equivalent.
for column in ["city", "service_type"]:
    before = df[column].astype("string")
    after = before.str.strip().str.title()
    changed = before.ne(after).fillna(False)
    df[column] = after
    transformation_log.append({
        "step": len(transformation_log) + 1, "variable": column,
        "action": "Strip edge spaces and use title case",
        "rationale": "Make formatting variants comparable",
        "rows_affected": int(changed.sum()), "reversible": "Yes, raw source retained",
        "notes": "No undocumented category aliases merged"
    })

numeric_columns = [
    "cases_received", "cases_resolved", "resolution_days", "satisfaction_score",
    "digital_cases", "in_person_cases", "repeat_contacts", "complaints_received", "staff_hours"
]
for column in numeric_columns:
    before = df[column].copy()
    converted = pd.to_numeric(before, errors="coerce")
    failed = before.notna() & converted.isna()
    df[column] = converted
    transformation_log.append({
        "step": len(transformation_log) + 1, "variable": column,
        "action": "Convert numeric representation",
        "rationale": "Expose unconvertible values and enable numeric checks",
        "rows_affected": int(failed.sum()), "reversible": "Yes, raw source retained",
        "notes": "Count is newly missing values, not all valid numeric conversions"
    })

before_month = df["report_month"].copy()
df["report_month"] = pd.to_datetime(before_month, format="%Y-%m-%d", errors="coerce")
failed_dates = before_month.notna() & df["report_month"].isna()
transformation_log.append({
    "step": len(transformation_log) + 1, "variable": "report_month",
    "action": "Parse YYYY-MM-DD dates", "rationale": "Expose invalid dates without guessing",
    "rows_affected": int(failed_dates.sum()), "reversible": "Yes, raw source retained",
    "notes": "Count is newly missing dates; invalid source values become NaT"
})

# A flag requests review. Keep questionable records in the master table.
df["flag_invalid_month"] = df["report_month"].isna()
df["flag_invalid_satisfaction"] = df["satisfaction_score"].notna() & ~df["satisfaction_score"].between(1, 5)
df["flag_negative_cases_received"] = df["cases_received"] < 0
df["flag_negative_complaints"] = df["complaints_received"] < 0
df["flag_zero_or_negative_staff_hours"] = df["staff_hours"] <= 0
df["flag_resolved_gt_received"] = df["cases_resolved"] > df["cases_received"]
channels_present = df[["cases_received", "digital_cases", "in_person_cases"]].notna().all(axis=1)
df["flag_channel_mismatch"] = channels_present & (
    df["digital_cases"] + df["in_person_cases"] != df["cases_received"]
)
df["flag_repeat_gt_received"] = df["repeat_contacts"] > df["cases_received"]
df["flag_repeated_id"] = df["record_id"].duplicated(keep=False)
known_cities = {"Copenhagen", "Aarhus", "Odense", "Aalborg"}
known_services = {"Housing", "Transport", "Employment", "Waste", "Citizen Services"}
df["flag_unrecognised_city"] = df["city"].notna() & ~df["city"].isin(known_cities)
df["flag_unrecognised_service"] = df["service_type"].notna() & ~df["service_type"].isin(known_services)
df["flag_missing_feedback"] = df["feedback"].fillna("").str.strip().eq("")
flag_columns = [column for column in df.columns if column.startswith("flag_")]
df["flag_any_quality_issue"] = df[flag_columns].any(axis=1)
for column in flag_columns:
    transformation_log.append({
        "step": len(transformation_log) + 1, "variable": column,
        "action": "Flag and retain", "rationale": "Make unresolved issues visible",
        "rows_affected": int(df[column].sum()), "reversible": "Yes, no record removed",
        "notes": "Counts may overlap; False does not establish complete validity"
    })

import pandas as pd  # Work with tables and missing values.
import nltk  # Tokenisation and English stop-word tools.
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# A fresh runtime needs these resources. Existing downloads are reused.
resources = [('punkt', 'tokenizers/punkt'), ('punkt_tab', 'tokenizers/punkt_tab'), ('stopwords', 'corpora/stopwords')]
for name, resource_path in resources:
    try:
        nltk.data.find(resource_path)
    except LookupError:
        nltk.download(name, quiet=True, raise_on_error=True)
# Preserve these negation words because removing them can reverse meaning.
stop_words = set(stopwords.words("english")) - {"not", "no", "nor"}

def prepare_feedback(text):
    if pd.isna(text):
        return []
    tokens = word_tokenize(str(text).lower())
    return [token for token in tokens
            if token.isalpha() and token not in stop_words]
# The original feedback stays unchanged; missingness has a separate flag.
df["feedback_tokens"] = df["feedback"].apply(prepare_feedback)
df["feedback_clean"] = df["feedback_tokens"].apply(lambda tokens: " ".join(tokens))
feedback_present = df["feedback"].notna() & df["feedback"].fillna("").str.strip().ne("")
transformation_log.append({
    "step": len(transformation_log) + 1, "variable": "feedback",
    "action": "Derive tokens and prepared text", "rationale": "Prepare a separate representation for counting",
    "rows_affected": int(feedback_present.sum()), "reversible": "Yes, original feedback retained",
    "notes": "Count is non-empty source feedback; empty token lists do not imply neutral sentiment"
})

log_df = pd.DataFrame(transformation_log)

# MAIN DEMONSTRATION
from pathlib import Path
# Save student copies in this runtime only. No repository data is overwritten.
output_dir = Path.cwd() / "tutorial_outputs"
output_dir.mkdir(parents=True, exist_ok=True)
clean_file = output_dir / "L06_student_cleaned_copy.csv"
log_file = output_dir / "L06_student_transformation_log.csv"
df.to_csv(clean_file, index=False)
log_df.to_csv(log_file, index=False)
print("Cleaned student copy:", clean_file)
print("Transformation log copy:", log_file)
print("Rows read back from saved copy:", len(pd.read_csv(clean_file)))
print("Log entries read back:", len(pd.read_csv(log_file)))
print("\nCleaning pipeline completed.")
print("Raw rows:", len(df_raw))
print("Cleaned master rows:", len(df))
print("Raw source preserved:", df["feedback"].equals(df_raw.loc[df.index, "feedback"]))
print("Quality flags retained:", bool(flag_columns))
print("Prepared text column created:", "feedback_clean" in df)
print("Transformation log created:", not log_df.empty)
print("Ready for Lecture 7 analysis: yes, with documented limitations.")
````
