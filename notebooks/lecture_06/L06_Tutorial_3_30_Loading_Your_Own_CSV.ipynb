## Example 3.30.1 — Ask the student to upload a CSV in Colab

Run the whole block in Google Colab. Choose one CSV from your computer. The code reads the file you selected, whatever its filename, and prints the dimensions, column names and first five rows. It does not assume any course-specific columns.

```python
import pandas as pd  # Read a CSV into a table called a DataFrame.
import io  # Let pandas read the uploaded file's bytes.
from google.colab import files  # Colab's upload dialogue; run this in Colab.

print("Choose ONE CSV file from your computer.")
uploaded = files.upload()  # Opens the file chooser and waits for your selection.

if len(uploaded) != 1:
    print("Please run this block again and choose exactly one CSV file.")
else:
    filename = next(iter(uploaded))  # Use the actual uploaded filename.
    source_df = pd.read_csv(io.BytesIO(uploaded[filename]), encoding="utf-8-sig")
    df = source_df.copy(deep=True)  # Work on a copy; preserve the loaded table.

    print("Loaded file:", filename)
    print("Rows:", df.shape[0], "| Columns:", df.shape[1])
    print("Column names:", df.columns.tolist())
    print("\nFirst five rows:")
    print(df.head(5).to_string(index=False))
```

**Expected interaction:** select one CSV in Colab. With the supplied practice file, the table has **5 rows and 4 columns**. Your own CSV will produce different values. The cell is intentionally unexecuted in the distributed notebook because it needs your file selection. A cancelled upload or incorrect path produces guidance; a wrongly formatted CSV may raise a parsing or encoding error. See the troubleshooting section.

## Example 3.30.2 — Read a file path inside Colab

First upload one CSV using Colab's Files panel. Use the file's Copy path command, then run this block and paste that path, for example /content/my_survey.csv. The block includes all imports and loading; it does not depend on Example 3.30.1. The file must exist in this Colab runtime.

## Example 3.30.4 — Understand a relative path with a small practice file

This self-contained demonstration creates a small synthetic CSV in a temporary practice folder, then reads it using a relative path. The relative path is interpreted from the stated base folder. In your own project, data/my_survey.csv is interpreted from Path.cwd(). The temporary demonstration file is removed automatically afterwards.

```python
import pandas as pd
from pathlib import Path
from tempfile import TemporaryDirectory  # A disposable practice folder.

# Everything needed for this example is created in this block.
with TemporaryDirectory() as practice_folder:
    base_folder = Path(practice_folder)
    data_folder = base_folder / "data"
    data_folder.mkdir()
    relative_path = Path("data") / "my_survey.csv"
    csv_path = base_folder / relative_path  # Join the base and relative path.
    csv_path.write_text("activity,minutes\nInterview,25\nWorkshop,40\n", encoding="utf-8")

    df = pd.read_csv(csv_path)
    print("Relative path:", relative_path.as_posix())
    print("File exists:", csv_path.is_file())
    print("Rows:", len(df), "| Columns:", len(df.columns))
    print(df.to_string(index=False))
```

## Example 3.30.6 — Recognise a semicolon CSV and decimal commas

CSV files can use different separators. A semicolon-delimited file may load as one wide text column if you assume commas. Here a tiny in-memory example makes the problem visible, then sep=';' and decimal=',' load it correctly. Select settings from the file's documented format; these settings are not suitable for every CSV.

```python
import pandas as pd
from io import StringIO  # A small CSV stored as text, for a repeatable demonstration.

csv_text = "activity;score\nInterview;4,5\nWorkshop;3,0\n"
wrong = pd.read_csv(StringIO(csv_text))
df = pd.read_csv(StringIO(csv_text), sep=";", decimal=",")
print("With the default comma separator:")
print("Column names:", wrong.columns.tolist())
print("\nWith sep=';' and decimal=',':")
print(df.to_string(index=False))
print("Score data type:", df["score"].dtype)
```

## Example 3.30.7 — Integrated example: upload, inspect and preserve your own data

Upload any small CSV and inspect its structure, data types, missing values and exact duplicate count. All inspection is column-independent. Change CSV_SEPARATOR, CSV_DECIMAL and CSV_ENCODING to match your file before running. Loading and counting do not automatically clean the data or justify deleting rows.

```python
import pandas as pd
import io
from google.colab import files

# Match these settings to your CSV. For a semicolon CSV, change the separator.
CSV_SEPARATOR = ","
CSV_DECIMAL = "."
CSV_ENCODING = "utf-8-sig"  # Handles standard UTF-8 with or without a BOM.

print("Choose ONE CSV file to inspect.")
uploaded = files.upload()
if len(uploaded) != 1:
    print("Run again and choose exactly one CSV file.")
else:
    filename = next(iter(uploaded))
    source_df = pd.read_csv(io.BytesIO(uploaded[filename]),
                            sep=CSV_SEPARATOR, decimal=CSV_DECIMAL,
                            encoding=CSV_ENCODING)
    df = source_df.copy(deep=True)
    print("Loaded:", filename)
    print("Rows:", len(df), "| Columns:", len(df.columns))
    print("\nFirst five rows:")
    print(df.head(5).to_string(index=False))
    print("\nColumn names:", df.columns.tolist())
    print("\nData types:")
    print(df.dtypes.to_string())
    print("\nMissing values by column:")
    print(df.isna().sum().to_string())
    print("\nExact duplicate rows:", df.duplicated().sum())
    print("Loaded table unchanged:", df.equals(source_df))
```

## Checks and troubleshooting
| What you see | What to check |
| --- | --- |
| `ModuleNotFoundError: google.colab` | The upload chooser is a Colab tool. Open this notebook in Colab, or use a local-path example in local Python. |
| File not found | Check which machine runs Python, the current folder, filename, extension and letter case. Paste a path without surrounding quotes. A relative path is resolved from the current working folder. |
| No file or several files selected | Rerun the upload cell and choose exactly one CSV. |
| One column containing semicolons | Inspect the file's format; try `sep=";"` when semicolons separate fields. |
| `UnicodeDecodeError` | UTF-8 could not decode the file. Confirm the encoding with the file's source; use that documented encoding. `utf-8-sig` handles a UTF-8 BOM, not every encoding. Avoid ignoring errors, which can lose text. |
| `ParserError` or unexpected columns | Check the delimiter, quoting, header row and whether this is actually a CSV rather than an Excel workbook or HTML page. Do not silently skip troublesome rows. |
| `EmptyDataError` | The file has no readable columns. Open it and check its content. |
| Missing values appear | Blank CSV fields are normally read as missing values; loading is not a decision to fill or delete them. |
| Earlier analysis raises `KeyError` | A new CSV may not have `feedback`, `service_type` or other course fields. Inspect `df.columns` and adapt the analysis. |
