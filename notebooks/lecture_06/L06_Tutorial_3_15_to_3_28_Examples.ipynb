## Example 3.27.6 — Build the cleaning function

**Website locator:** Example 3.27.6

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.
import nltk  # Imports the Natural Language Toolkit used for introductory text analysis.
from nltk.tokenize import word_tokenize  # Imports the word tokenizer.

def prepare_feedback(text):
    if pd.isna(text):
        return []

    tokens = word_tokenize(str(text).lower())

    cleaned_tokens = []

    for token in tokens:
        if token.isalpha() and token not in stop_words:
            cleaned_tokens.append(token)

    return cleaned_tokens
```

## Example 3.28.5 — Standardise category spacing and capitalisation

**Website locator:** Example 3.28.5

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.
import re  # Imports regular-expression tools for transparent text cleaning.

for column in ["city", "service_type"]:
    before = df[column].copy()

    df[column] = (
        df[column]
        .str.strip()
        .str.title()
    )

    rows_changed = (
        before.fillna("<MISSING>")
        != df[column].fillna("<MISSING>")
    ).sum()

    transformation_log.append({
        "step": len(transformation_log) + 1,
        "column": column,
        "issue": "Inconsistent spacing/capitalisation",
        "action": "Applied str.strip() and str.title()",
        "reason": "Reduce artificial category fragmentation",
        "rows_affected": int(rows_changed),
        "possible_effect": "Category counts may merge previously separated spellings"
    })
```

## Example 3.28.6 — Convert intended numeric columns

**Website locator:** Example 3.28.6

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.
import re  # Imports regular-expression tools for transparent text cleaning.

numeric_columns = [
    "cases_received",
    "cases_resolved",
    "resolution_days",
    "satisfaction_score",
    "digital_cases",
    "in_person_cases",
    "repeat_contacts",
    "complaints_received",
    "staff_hours"
]

for column in numeric_columns:
    before = df[column].copy()

    converted = pd.to_numeric(before, errors="coerce")

    newly_missing = (
        before.notna()
        & converted.isna()
    ).sum()

    df[column] = converted

    transformation_log.append({
        "step": len(transformation_log) + 1,
        "column": column,
        "issue": "Column expected to be numeric",
        "action": "Converted with pd.to_numeric(errors='coerce')",
        "reason": "Enable valid numeric calculations and expose unconvertible values",
        "rows_affected": int(newly_missing),
        "possible_effect": "Unconvertible source values become missing and are excluded from many calculations"
    })
```

## Example 3.28.7 — Convert the month column

**Website locator:** Example 3.28.7

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.
import re  # Imports regular-expression tools for transparent text cleaning.

report_month_before = df["report_month"].copy()

df["report_month"] = pd.to_datetime(
    df["report_month"],
    errors="coerce"
)

new_invalid_dates = (
    report_month_before.notna()
    & df["report_month"].isna()
).sum()

transformation_log.append({
    "step": len(transformation_log) + 1,
    "column": "report_month",
    "issue": "Month values stored as text / possible invalid date",
    "action": "Converted with pd.to_datetime(errors='coerce')",
    "reason": "Enable chronological analysis and expose invalid dates",
    "rows_affected": int(new_invalid_dates),
    "possible_effect": "Invalid dates become NaT and cannot be used in time-based analysis"
})
```

## Example 3.28.8 — Create quality flags

**Website locator:** Example 3.28.8

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.

df["flag_invalid_satisfaction"] = (
    (df["satisfaction_score"] < 1)
    | (df["satisfaction_score"] > 5)
)

df["flag_negative_cases_received"] = (
    df["cases_received"] < 0
)

df["flag_negative_complaints"] = (
    df["complaints_received"] < 0
)

df["flag_zero_or_negative_staff_hours"] = (
    df["staff_hours"] <= 0
)

df["flag_resolved_gt_received"] = (
    df["cases_resolved"] > df["cases_received"]
)

required_channels_available = df[[
    "cases_received",
    "digital_cases",
    "in_person_cases"
]].notna().all(axis=1)

df["flag_channel_mismatch"] = (
    required_channels_available
    & (
        df["digital_cases"]
        + df["in_person_cases"]
        != df["cases_received"]
    )
)

df["flag_repeat_gt_received"] = (
    df["repeat_contacts"] > df["cases_received"]
)

df["flag_repeated_id"] = (
    df["record_id"].duplicated(keep=False)
)
```

## Example 3.28.10 — Prepare text

**Website locator:** Example 3.28.10

```python
# Imports are repeated intentionally so this code block is easier to copy into a fresh notebook cell.
import pandas as pd  # Imports pandas for working with table-shaped data.
import nltk  # Imports the Natural Language Toolkit used for introductory text analysis.
from nltk.tokenize import word_tokenize  # Imports the word tokenizer.
from nltk.corpus import stopwords  # Imports the English stop-word list.

stop_words = set(stopwords.words("english"))
stop_words.difference_update({"not", "no", "nor"})


def prepare_feedback(text):
    if pd.isna(text):
        return []

    tokens = word_tokenize(str(text).lower())

    return [
        token
        for token in tokens
        if token.isalpha()
        and token not in stop_words
    ]


df["feedback_tokens"] = df["feedback"].apply(prepare_feedback)

df["feedback_clean"] = df["feedback_tokens"].apply(
    lambda tokens: " ".join(tokens)
)
```
