## Example 3.24.11 — Integrated cleaning, inspection and transformation log

**Website locator:** Example 3.24.11

**Situation:** A researcher has a four-row synthetic service table. City names contain accidental spaces, one staff-hours value is written as `unknown`, one satisfaction score is outside the allowed 1–5 range and another score is missing.

**What this example is for:** Combine the four responses from Tutorial 3.24 in one runnable example: correct formatting, convert a value, flag and retain a questionable record, and exclude rows from one specific analysis. Count affected rows and record a reason for each decision.

**Learning outcome:** Explain and document how a cleaning choice changes the data available for analysis. This supports the module objectives on writing simple scripts, handling data and assessing data quality and reliability.

**Run it on its own:** This cell creates its own four-row table, so it needs only pandas and no CSV download. It uses `example_raw` and `example_df` to keep the recurring course table `df` available for the following tutorials.

**Read the example in this order:**
1. Create the synthetic data and a separate working copy.
2. Remove edge spaces from city labels and count the changed rows.
3. Convert staff hours to numbers and count values that become missing.
4. Flag scores outside 1–5, while counting missing scores separately.
5. Create a separate table for the satisfaction analysis.
6. Build the transformation log and print the original table, working table, analysis subset and log together.

**Before you run it:** Predict how many city labels will change, how many staff-hours values will become missing, how many scores will be flagged and how many rows the satisfaction analysis will exclude.


```python
import pandas as pd

# Small example dataset, so this code runs on its own.
example_raw = pd.DataFrame({
    "city": [" Copenhagen ", "Aarhus", " Aalborg", "Odense "],
    "staff_hours": ["40", "unknown", "25", "30"],
    "satisfaction_score": [4, 7, 3, None]
})

# Work on a copy to preserve the original data.
# example_df keeps this small demonstration separate from the course table df.
example_df = example_raw.copy()


# 1. CORRECT: remove spaces around city names.
city_before = example_df["city"].copy()
example_df["city"] = example_df["city"].str.strip()

# Count rows whose city name changed.
city_changed = (
    city_before.fillna("") != example_df["city"].fillna("")
).sum()


# 2. CONVERT: turn staff hours into numbers.
hours_before = example_df["staff_hours"].copy()
example_df["staff_hours"] = pd.to_numeric(
    example_df["staff_hours"],
    errors="coerce"  # Unconvertible text becomes NaN (missing).
)

# Count values that became missing during conversion.
hours_failed = (
    hours_before.notna() & example_df["staff_hours"].isna()
).sum()


# 3. FLAG AND RETAIN: identify scores outside the allowed range.
# Scores should be between 1 and 5. Keep questionable rows for review.
example_df["flag_invalid_satisfaction"] = (
    (example_df["satisfaction_score"] < 1)
    | (example_df["satisfaction_score"] > 5)
)

# Missing scores are checked separately from out-of-range scores.
invalid_scores = example_df["flag_invalid_satisfaction"].sum()
missing_scores = example_df["satisfaction_score"].isna().sum()


# 4. EXCLUDE FOR ONE ANALYSIS: select valid satisfaction scores.
# This creates a separate table. It does not delete rows from example_df.
example_valid_satisfaction = example_df[
    example_df["satisfaction_score"].between(1, 5)
].copy()

scores_excluded = len(example_df) - len(example_valid_satisfaction)


# Record what happened, how many rows it affected and why.
example_log = pd.DataFrame([
    {
        "action": "Removed spaces from city names",
        "rows_affected": int(city_changed),
        "reason": "Avoid separate categories caused by accidental spaces"
    },
    {
        "action": "Converted staff hours to numeric",
        "rows_affected": int(hours_failed),
        "reason": "Expose unconvertible values as missing"
    },
    {
        "action": "Flagged out-of-range satisfaction scores",
        "rows_affected": int(invalid_scores),
        "reason": "Retain questionable values for source verification"
    },
    {
        "action": "Excluded scores from satisfaction analysis",
        "rows_affected": int(scores_excluded),
        "reason": "This analysis requires a score between 1 and 5"
    }
])


# Print all results together.
print("ORIGINAL DATA")
print(example_raw.to_string(index=False))

print("\nWORKING DATA AFTER CLEANING AND INSPECTION")
print(example_df.to_string(index=False))

print("\nVALID ROWS FOR SATISFACTION ANALYSIS")
print(example_valid_satisfaction.to_string(index=False))

print("\nMissing satisfaction scores:", missing_scores)

print("\nTRANSFORMATION LOG")
print(example_log.to_string(index=False))
```

### Expected results and interpretation

| Action or check | Expected result | What it means |
|---|---:|---|
| City labels corrected | 3 rows | Copenhagen, Aalborg and Odense lose accidental edge spaces. |
| Staff-hours values becoming missing | 1 value | `unknown` becomes `NaN`. It does not become zero. |
| Out-of-range satisfaction scores flagged | 1 row | Aarhus has a score of 7. The row remains in `example_df`. |
| Missing satisfaction scores | 1 value | Odense has no recorded score. Missing is checked separately. |
| Rows excluded from the satisfaction subset | 2 rows | The score of 7 and the missing score cannot enter this analysis. |
| Rows retained for the satisfaction analysis | 2 rows | Copenhagen and Aalborg have scores between 1 and 5. |

The transformation log prints the affected-row counts **3, 1, 1 and 2** for its four actions. The working table still contains all four rows, and the raw table stays unchanged. These counts refer to different operations and can overlap; adding them does not give a count of unique problem rows.

**Interpret carefully:** A `False` value in `flag_invalid_satisfaction` only means the score was not flagged as outside the range. A missing score also receives `False` here, so it still needs the separate missing-value check. Successful numeric conversion does not prove that a value is accurate, and a score of 7 needs source verification before correction. The satisfaction subset represents only rows with an available in-range score.
