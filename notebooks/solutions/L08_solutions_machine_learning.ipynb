## Steps

1. Split Iris into training and test data with a fixed random state and stratification.
2. Train a shallow decision tree and generate test predictions.
3. Report accuracy and a confusion matrix.
4. Write a model record with at least three limits.


```python
from sklearn.datasets import load_iris  # Imports the built-in labelled Iris dataset.
from sklearn.model_selection import train_test_split  # Imports a reproducible splitting function.
from sklearn.tree import DecisionTreeClassifier  # Imports a simple tree classifier.
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay  # Imports two evaluation tools.
import matplotlib.pyplot as plt  # Imports plotting tools for the confusion matrix.
iris_bundle = load_iris(as_frame=True)  # Loads features, labels, and descriptive metadata.
features = iris_bundle.data.copy()  # Copies measured features into their own DataFrame.
labels = iris_bundle.target.copy()  # Copies species labels into their own Series.
training_features, test_features, training_labels, test_labels = train_test_split(features, labels, test_size=0.25, random_state=42, stratify=labels)  # Creates a reproducible stratified holdout set.
model = DecisionTreeClassifier(max_depth=3, random_state=42)  # Creates a bounded reproducible classifier.
model.fit(training_features, training_labels)  # Learns only from the training rows.
predictions = model.predict(test_features)  # Generates labels for the held-out feature rows.
accuracy = accuracy_score(test_labels, predictions)  # Calculates the proportion of correct test predictions.
print(f"Test accuracy: {accuracy:.3f}")  # Displays the holdout accuracy.
ConfusionMatrixDisplay.from_predictions(test_labels, predictions, display_labels=iris_bundle.target_names, cmap="Blues")  # Draws class-specific observed and predicted counts.
plt.title("Iris test-set confusion matrix")  # Labels the figure with its evaluation scope.
plt.tight_layout()  # Prevents class labels from being cut off.
plt.show()  # Displays the evaluation figure.
model_record = {"task": "Iris species classification", "training_rows": len(training_features), "test_rows": len(test_features), "metric": "test accuracy and confusion matrix", "limits": ["small historical teaching dataset", "one train/test split", "performance may differ for new populations"]}  # Documents the pipeline and three limits.
print(model_record)  # Displays the complete model record.
```

## Checks

The held-out test rows must not be used in `fit()`. The model record must not claim general performance beyond the evaluated Iris split.
