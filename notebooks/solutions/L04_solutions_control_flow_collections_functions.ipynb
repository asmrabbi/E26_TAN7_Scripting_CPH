## Steps

1. Write a function that returns problems for an invalid satisfaction score.
2. Test the function with 1.0, 5.0, 0.9, 5.1, and None.
3. Loop over a list of records and count how many require review.
4. Explain why the two exact boundary values are important tests.


```python
def validate_score(score):  # Defines a reusable satisfaction-score validation function.
    """Return a list describing definite problems with one score."""  # Documents the function output.
    problems = []  # Creates an empty list for zero or more problem messages.
    if score is None:  # Checks whether the score is missing.
        problems.append("score is missing")  # Records the missing-value problem.
        return problems  # Stops safely because numeric comparisons would fail for None.
    if score < 1 or score > 5:  # Checks both ends of the stated one-to-five scale.
        problems.append("score is outside 1–5")  # Records the range problem.
    return problems  # Returns the complete result to the caller.
test_scores = [1.0, 5.0, 0.9, 5.1, None]  # Stores normal, boundary, invalid, and missing test cases.
review_count = 0  # Starts the number of problematic scores at zero.
for test_score in test_scores:  # Applies the same function to every test case.
    score_problems = validate_score(test_score)  # Collects any problems for the current value.
    if score_problems:  # Checks whether the result contains at least one problem.
        review_count = review_count + 1  # Increases the review counter.
    print(test_score, score_problems)  # Displays the input and observed result together.
print(f"Scores requiring review: {review_count}")  # Displays the final count of invalid or missing cases.
```

## Checks

1.0 and 5.0 are valid boundaries; 0.9, 5.1, and None require review, so the expected count is 3.
