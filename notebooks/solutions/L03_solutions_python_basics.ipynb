## Steps

1. Create variables for one service record and print a readable one-line summary.
2. Calculate unresolved cases and a resolution percentage.
3. Convert a score stored as text into a number and add 0.1.
4. Run two normal inputs and one edge case, then record what happened.


```python
record_id = 2001  # Stores the identifier supplied for the exercise.
service_type = "Transport"  # Stores the service category supplied for the exercise.
cases_received = 85  # Stores the received count supplied for the exercise.
cases_resolved = 80  # Stores the resolved count supplied for the exercise.
score_text = "4.0"  # Stores a satisfaction score as text so conversion is visible.
unresolved_cases = cases_received - cases_resolved  # Calculates how many received cases remain unresolved.
resolution_percentage = cases_resolved / cases_received * 100  # Calculates the resolved share as a percentage.
score_number = float(score_text)  # Converts the text score into a decimal number.
print(f"Record {record_id}: {service_type}; unresolved={unresolved_cases}; resolved={resolution_percentage:.1f}%; adjusted score={score_number + 0.1:.1f}")  # Displays all calculated values in one readable line.
```
