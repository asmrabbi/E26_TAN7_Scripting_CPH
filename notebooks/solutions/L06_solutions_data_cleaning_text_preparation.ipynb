## Steps

1. Preserve the raw DataFrame and make a working copy.
2. Standardise city labels, convert numeric columns, and remove exact duplicates.
3. Flag invalid case counts and scores instead of silently deleting them.
4. Tokenise feedback and print a complete transformation log.


```python
from pathlib import Path  # Imports Path so the notebook can find a local course file when available.
import pandas as pd  # Imports pandas for reading and working with table-shaped data.
remote_data_url = "https://raw.githubusercontent.com/asmrabbi/E26_TAN7_Scripting_CPH/main/data/monthly_service_report.csv"  # Stores the public GitHub address used by Google Colab.
local_data_candidates = [Path("data/monthly_service_report.csv"), Path("../../data/monthly_service_report.csv")]  # Lists possible local paths used during validation.
data_source = next((path for path in local_data_candidates if path.exists()), remote_data_url)  # Chooses a local file when present and otherwise uses GitHub.
raw_exercise_data = pd.read_csv(data_source)  # Reads the CSV file into a pandas DataFrame.
print(raw_exercise_data.head())  # Prints a small preview so we can confirm that loading worked.
import re  # Imports regular-expression tools for transparent word tokenisation.
working_data = raw_exercise_data.copy()  # Creates a working copy while preserving the loaded source.
transformation_log = []  # Creates a list for documenting each justified change.
working_data["city"] = working_data["city"].astype("string").str.strip().str.lower().replace({"koebenhavn": "copenhagen"}).str.title()  # Standardises known city variants.
transformation_log.append("Standardised whitespace, case, and documented city spelling variants.")  # Records the category transformation.
for column_name in ["cases_received", "cases_resolved", "resolution_days", "satisfaction_score"]:  # Visits every expected numeric column.
    working_data[column_name] = pd.to_numeric(working_data[column_name], errors="coerce")  # Converts valid numbers and exposes failures as missing.
transformation_log.append("Converted expected numeric columns; invalid text became missing for review.")  # Records the numeric conversion rule.
rows_before = len(working_data)  # Stores the number of rows before removing confirmed duplicates.
working_data = working_data.drop_duplicates().copy()  # Removes only exact duplicate rows.
transformation_log.append(f"Removed {rows_before - len(working_data)} exact duplicate row(s).")  # Records the precise row-count effect.
working_data["case_count_problem"] = (working_data["cases_received"] < 0) | (working_data["cases_resolved"] < 0) | (working_data["cases_resolved"] > working_data["cases_received"])  # Flags case-count rule failures.
working_data["score_problem"] = ~working_data["satisfaction_score"].between(1, 5) & working_data["satisfaction_score"].notna()  # Flags non-missing scores outside the scale.
feedback_text = working_data["feedback"].dropna().astype(str).str.cat(sep=" ").lower()  # Combines available feedback into lowercase text.
feedback_tokens = re.findall(r"[a-z]+", feedback_text)  # Extracts alphabetic tokens with a visible rule.
print(feedback_tokens[:20])  # Displays a bounded token sample.
for log_entry in transformation_log:  # Visits each transformation in the order performed.
    print(log_entry)  # Displays the current transformation entry.
print(working_data.isna().sum())  # Displays remaining missing values rather than hiding them.
print(working_data[["record_id", "case_count_problem", "score_problem"]])  # Displays the record-level audit flags.
```
