## Steps

1. Report the dataset shape, column names, and inferred data types.
2. Count missing values and exact duplicates.
3. List the raw labels in city and service_type.
4. Write a four-field data note: source, unit, size, and visible concerns.


```python
from pathlib import Path  # Imports Path so the notebook can find a local course file when available.
import pandas as pd  # Imports pandas for reading and working with table-shaped data.
remote_data_url = "https://raw.githubusercontent.com/asmrabbi/E26_TAN7_Scripting_CPH/main/data/monthly_service_report.csv"  # Stores the public GitHub address used by Google Colab.
local_data_candidates = [Path("data/monthly_service_report.csv"), Path("../../data/monthly_service_report.csv")]  # Lists possible local paths used during validation.
data_source = next((path for path in local_data_candidates if path.exists()), remote_data_url)  # Chooses a local file when present and otherwise uses GitHub.
exercise_data = pd.read_csv(data_source)  # Reads the CSV file into a pandas DataFrame.
print(exercise_data.head())  # Prints a small preview so we can confirm that loading worked.
print(exercise_data.shape)  # Displays the number of rows and columns.
print(exercise_data.columns.tolist())  # Displays every column name.
print(exercise_data.dtypes)  # Displays pandas inferred types for each column.
print(exercise_data.isna().sum())  # Displays the missing-value count for every column.
print(f"Exact duplicates: {exercise_data.duplicated().sum()}")  # Displays the number of repeated whole rows.
print(sorted(exercise_data["city"].dropna().astype(str).unique().tolist()))  # Displays distinct raw city labels.
print(sorted(exercise_data["service_type"].dropna().astype(str).unique().tolist()))  # Displays distinct raw service labels.
data_note = {"source": "synthetic AAU teaching dataset", "unit": "service category by city and month", "size": exercise_data.shape, "visible_concerns": ["missing values", "duplicate rows", "inconsistent labels", "mixed numeric formats"]}  # Documents source, unit, size, and observed concerns.
print(data_note)  # Displays the completed data note.
```
