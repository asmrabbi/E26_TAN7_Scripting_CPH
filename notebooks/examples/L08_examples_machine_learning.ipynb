## Goal

Train and evaluate a simple classifier with the Iris dataset, then document what the result does and does not support.


## Setup

The Iris dataset is bundled with scikit-learn, so no upload or private data is required.


## Steps

### 1. Load labelled examples


```python
import pandas as pd  # Imports pandas for readable tables.
from sklearn.datasets import load_iris  # Imports the built-in Iris teaching dataset.
iris_bundle = load_iris(as_frame=True)  # Loads features, labels, names, and source description together.
features = iris_bundle.data.copy()  # Copies the four measured flower features into a DataFrame.
labels = iris_bundle.target.copy()  # Copies the numeric species labels into a separate Series.
print(features.head())  # Displays a bounded preview of the feature values.
print(labels.value_counts().sort_index())  # Displays the number of examples in each label class.
```

### 2. Create separate training and test data


```python
from sklearn.model_selection import train_test_split  # Imports a function for making a reproducible data split.
training_features, test_features, training_labels, test_labels = train_test_split(features, labels, test_size=0.25, random_state=42, stratify=labels)  # Reserves 25% for one final evaluation while preserving class proportions.
print(training_features.shape)  # Displays the number of training rows and feature columns.
print(test_features.shape)  # Displays the number of test rows and feature columns.
```

### 3. Train a simple decision tree


```python
from sklearn.tree import DecisionTreeClassifier  # Imports an interpretable tree-based classifier.
classifier = DecisionTreeClassifier(max_depth=3, random_state=42)  # Creates a small reproducible model and limits its depth.
classifier.fit(training_features, training_labels)  # Learns decision rules using only the training data.
test_predictions = classifier.predict(test_features)  # Applies the learned rules to unseen test features.
print(test_predictions[:10])  # Displays a bounded sample of predicted class numbers.
```

### 4. Evaluate predictions


```python
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay, classification_report  # Imports three complementary evaluation tools.
import matplotlib.pyplot as plt  # Imports plotting tools for the confusion matrix.
test_accuracy = accuracy_score(test_labels, test_predictions)  # Calculates the proportion of correct test predictions.
print(f"Test accuracy: {test_accuracy:.3f}")  # Displays the result to three decimal places.
print(classification_report(test_labels, test_predictions, target_names=iris_bundle.target_names, zero_division=0))  # Displays per-class precision, recall, and F1 values.
ConfusionMatrixDisplay.from_predictions(test_labels, test_predictions, display_labels=iris_bundle.target_names, cmap="Blues")  # Plots observed labels against predicted labels.
plt.title("Iris test-set confusion matrix")  # Labels the figure with its evaluation scope.
plt.tight_layout()  # Adjusts spacing so labels remain visible.
plt.show()  # Displays the confusion matrix.
```

### 5. Inspect individual predictions


```python
prediction_review = test_features.copy()  # Copies the test features so evaluation does not alter the source split.
prediction_review["observed_species"] = [iris_bundle.target_names[label] for label in test_labels]  # Converts observed class numbers into species names.
prediction_review["predicted_species"] = [iris_bundle.target_names[label] for label in test_predictions]  # Converts predicted class numbers into species names.
prediction_review["correct"] = prediction_review["observed_species"] == prediction_review["predicted_species"]  # Marks whether each individual prediction was correct.
print(prediction_review.head(10))  # Displays a bounded audit table of features and outcomes.
```

### 6. Record the model and its limits


```python
model_record = {"task": "classify Iris species", "features": features.columns.tolist(), "label": "species", "training_rows": len(training_features), "test_rows": len(test_features), "metric": "accuracy plus per-class precision and recall", "known_limits": ["small historical teaching dataset", "test result is not proof of performance on new populations", "flower measurements do not justify human-focused claims"]}  # Documents the complete teaching pipeline and important limits.
for record_key, record_value in model_record.items():  # Visits each labelled model-record field.
    print(f"{record_key}: {record_value}")  # Displays the model record in a readable order.
```

## Checks

The test set is not used during training. The confusion matrix and per-class metrics show more than accuracy alone. Results describe this split of Iris, not all future data.
