## Practice 2.3.1 — Estimate complete workshop groups

**Case question:** How many complete groups of 24 can be formed from 137 registrations, and how many people remain?

**Code logic**

1. Store the registration count and group capacity.
2. Use floor division for complete groups.
3. Use remainder for people left over.
4. Display both results.

```python
registrations = 137  # Stores the number of fictional event registrations.
group_capacity = 24  # Stores the maximum number of people in one complete workshop group.
complete_groups = registrations // group_capacity  # Calculates the number of full groups that can be formed.
people_remaining = registrations % group_capacity  # Calculates how many people remain outside the full groups.
print("Complete groups:", complete_groups)  # Displays the number of full workshop groups.
print("People remaining:", people_remaining)  # Displays the number still needing a placement.
```

## Practice 2.6.1 — Capture a real conversion error safely

**Case question:** How can a learner inspect the exact ValueError raised by invalid numerical text without stopping the whole practice notebook?

**Code logic**

1. Store invalid numerical text.
2. Attempt the conversion inside a focused try block.
3. Catch ValueError only.
4. Display the error type and message.

```python
invalid_count_text = "twenty"  # Stores text that cannot be converted directly to an integer.
try:  # Begins the smallest block that may raise the expected conversion error.
    converted_count = int(invalid_count_text)  # Attempts to convert the invalid text into a whole number.
    print(converted_count)  # Displays the converted value only if the conversion unexpectedly succeeds.
except ValueError as error:  # Catches the specific conversion error so later notebook cells can still run.
    print(type(error).__name__)  # Displays the official Python error type.
    print(error)  # Displays Python's explanatory error message.
```

## Practice 2.7.1 — Create a transparent indicator statement

**Case question:** How can a script calculate an indicator while keeping its source counts, formula and limitation visible?

**Code logic**

1. Store the two source counts.
2. Calculate the indicator.
3. Create a sentence that includes numerator and denominator.
4. State a limitation beside the output.

```python
responses_received = 184  # Stores the numerator reported by the fictional consultation.
people_invited = 500  # Stores the denominator used by the response-rate calculation.
response_rate = (responses_received / people_invited) * 100  # Calculates the response share using the documented denominator.
indicator_statement = f"Response rate: {responses_received}/{people_invited} ({response_rate:.1f}%)."  # Builds a sentence that preserves the source counts.
limitation_statement = "This rate does not show who was unable or unwilling to respond."  # Records one interpretation limit that the calculation cannot resolve.
print(indicator_statement)  # Displays the transparent numerical statement.
print(limitation_statement)  # Displays the accompanying critical limitation.
```
