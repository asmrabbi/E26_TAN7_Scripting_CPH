## Exercise 2.3.1 — Expressions, arithmetic and precedence

**Case question:** A team needs both a completion rate and the number of full review batches required for its remaining records.

**Code logic**

1. Calculate the completed count.
2. Calculate the completion percentage with an explicit denominator.
3. Calculate complete groups and the remainder for review batches.
4. Display all results with units.

```python
total_records = 250  # Stores the complete size of the record collection.
missing_records = 18  # Stores the records that are not complete.
complete_records = total_records - missing_records  # Calculates the complete part of the collection.
completion_rate = (complete_records / total_records) * 100  # Calculates the percentage using the total as denominator.
review_batch_size = 25  # Stores the number of records that fit in one complete batch.
complete_batches = missing_records // review_batch_size  # Calculates how many full review batches can be made.
remaining_records = missing_records % review_batch_size  # Calculates how many records remain after the full batches.
print(f"Completion rate: {completion_rate:.1f}%")  # Displays the calculated rate with one decimal place.
print("Complete review batches:", complete_batches)  # Displays the number of full batches.
print("Records left over:", remaining_records)  # Displays the remainder after grouping.
```

### Exercise 2.7.2 — Municipal consultation intake

A municipality received 640 consultation responses. Eighteen responses lack consent information, seven are duplicates, and the file contains 10 of the 11 expected columns. Produce a transparent intake summary while stating that the two exclusion categories are assumed not to overlap.

**Code plan**

1. Store the expected structure and all source counts.
2. Calculate retained responses and the retention percentage.
3. Calculate the column difference and two Boolean flags.
4. Display a labelled report and inspect at least two types.

```python
# Exercise 2.7.2 — Municipal consultation intake starter
# Write the solution in this cell after completing the code plan.
# Step 1: Store the expected structure and all source counts.
# Step 2: Calculate retained responses and the retention percentage.
# Step 3: Calculate the column difference and two Boolean flags.
# Step 4: Display a labelled report and inspect at least two types.
```

#### Fully commented solution — Exercise 2.7.2 — Municipal consultation intake

```python
# Exercise 2.7.2 solution: categories are assumed not to overlap.
EXPECTED_COLUMNS = 11  # Stores the value needed by later steps.
responses_received = 640  # Stores the value needed by later steps.
missing_consent = 18  # Stores the value needed by later steps.
duplicate_responses = 7  # Stores the value needed by later steps.
received_columns = 10  # Stores the value needed by later steps.
excluded_responses = missing_consent + duplicate_responses  # Stores the value needed by later steps.
retained_responses = responses_received - excluded_responses  # Stores the value needed by later steps.
retention_rate = retained_responses / responses_received * 100  # Stores the value needed by later steps.
column_difference = EXPECTED_COLUMNS - received_columns  # Stores the value needed by later steps.
has_exclusions = excluded_responses > 0  # Stores the value needed by later steps.
structure_matches_expectation = received_columns == EXPECTED_COLUMNS  # Runs this documented step in the solution.
print("Retained responses:", retained_responses)  # Displays a labelled result for inspection.
print(f"Retention rate: {retention_rate:.1f}%")  # Displays a labelled result for inspection.
print("Column difference:", column_difference)  # Displays a labelled result for inspection.
print("Has exclusions:", has_exclusions)  # Displays a labelled result for inspection.
print("Structure matches expectation:", structure_matches_expectation)  # Displays a labelled result for inspection.
print("Count type:", type(retained_responses))  # Displays a labelled result for inspection.
print("Rate type:", type(retention_rate))  # Displays a labelled result for inspection.
```

### Exercise 2.7.4 — Communicating a societal indicator

A public report states that at least 75 per cent of surveyed residents should report service access. In the sample, 596 of 800 residents report access. Calculate the indicator and keep the project threshold and its limitations visible.

**Code plan**

1. Store the surveyed count, access count and stated threshold.
2. Calculate the access percentage and percentage-point difference.
3. Use an at-least comparison for the threshold boundary.
4. Display the results and a limitation note.

```python
# Exercise 2.7.4 — Communicating a societal indicator starter
# Write the solution in this cell after completing the code plan.
# Step 1: Store the surveyed count, access count and stated threshold.
# Step 2: Calculate the access percentage and percentage-point difference.
# Step 3: Use an at-least comparison for the threshold boundary.
# Step 4: Display the results and a limitation note.
```

#### Fully commented solution — Exercise 2.7.4 — Communicating a societal indicator

```python
# Exercise 2.7.4 solution: the threshold is a stated project choice, not a Python fact.
residents_surveyed = 800  # Stores the value needed by later steps.
residents_reporting_access = 596  # Stores the value needed by later steps.
stated_threshold_percent = 75.0  # Stores the value needed by later steps.
reported_access_rate = residents_reporting_access / residents_surveyed * 100  # Stores the value needed by later steps.
difference_from_threshold = reported_access_rate - stated_threshold_percent  # Stores the value needed by later steps.
meets_stated_threshold = reported_access_rate >= stated_threshold_percent  # Runs this documented step in the solution.
limitation_note = "This percentage does not show whose experiences are absent or whether access was defined consistently."  # Stores the value needed by later steps.
print(f"Reported access rate: {reported_access_rate:.1f}%")  # Displays a labelled result for inspection.
print(f"Difference from threshold: {difference_from_threshold:.1f} percentage points")  # Displays a labelled result for inspection.
print("Meets stated threshold:", meets_stated_threshold)  # Displays a labelled result for inspection.
print("Limitation:", limitation_note)  # Displays a labelled result for inspection.
```
