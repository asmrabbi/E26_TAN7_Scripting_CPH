```python
# Exercise 2.7.9 starter: Compare two denominators in a public-service survey
# Write your own solution below after completing the numbered task.
# Step 1: Preserve the contacted, completed-answer and reported-access counts as separate source values.
# Step 2: Calculate the response rate using contacted residents as the denominator.
# Step 3: Calculate the reported-access rate using completed answers as the denominator and compare it with the stated threshold.
# Step 4: Display both percentages, their denominators and the threshold Boolean in a readable evidence statement.
# Step 5: Explain how using all contacted residents as the access-rate denominator would answer a different question and could mislead readers.
```

```python
residents_contacted = 950  # Stores the source or calculated value for later use.
completed_answers = 812  # Stores the source or calculated value for later use.
respondents_reporting_access = 603  # Stores the source or calculated value for later use.
stated_access_threshold = 70.0  # Stores the source or calculated value for later use.
response_rate = completed_answers / residents_contacted * 100  # Stores the source or calculated value for later use.
reported_access_rate = respondents_reporting_access / completed_answers * 100  # Stores the source or calculated value for later use.
meets_stated_access_threshold = reported_access_rate >= stated_access_threshold  # Runs this documented step in the solution.
response_statement = f"Response rate: {completed_answers} of {residents_contacted} contacted residents ({response_rate:.1f}%)."  # Stores the source or calculated value for later use.
access_statement = f"Reported access: {respondents_reporting_access} of {completed_answers} respondents ({reported_access_rate:.1f}%)."  # Stores the source or calculated value for later use.
denominator_note = "The response rate uses contacted residents; the access rate uses completed answers."  # Stores the source or calculated value for later use.
print(response_statement)  # Displays a labelled value so the result can be inspected.
print(access_statement)  # Displays a labelled value so the result can be inspected.
print("Meets stated access threshold:", meets_stated_access_threshold)  # Displays a labelled value so the result can be inspected.
print("Denominators:", denominator_note)  # Displays a labelled value so the result can be inspected.
```
