## Exercise 2.3.1 — Expressions, arithmetic and precedence

**Case question:** A team needs both a completion rate and the number of full review batches required for its remaining records.

**Code logic**

1. Calculate the completed count.
2. Calculate the completion percentage with an explicit denominator.
3. Calculate complete groups and the remainder for review batches.
4. Display all results with units.

```python
total_records = 250  # Stores the complete size of the record collection.
missing_records = 18  # Stores the records that are not complete.
complete_records = total_records - missing_records  # Calculates the complete part of the collection.
completion_rate = (complete_records / total_records) * 100  # Calculates the percentage using the total as denominator.
review_batch_size = 25  # Stores the number of records that fit in one complete batch.
complete_batches = missing_records // review_batch_size  # Calculates how many full review batches can be made.
remaining_records = missing_records % review_batch_size  # Calculates how many records remain after the full batches.
print(f"Completion rate: {completion_rate:.1f}%")  # Displays the calculated rate with one decimal place.
print("Complete review batches:", complete_batches)  # Displays the number of full batches.
print("Records left over:", remaining_records)  # Displays the remainder after grouping.
```

## Exercise 2.5.1 — Type conversion and simulated user input

**Case question:** A form sends a whole-number count and a place name as text. The notebook must preserve the raw text and create useful working values.

**Code logic**

1. Store the raw form values as strings.
2. Convert only the value that represents a count.
3. Standardise the place name for a draft file label.
4. Display both the raw and converted values for traceability.

```python
raw_response_count = "240"  # Simulates a numerical value received from a text field.
raw_place_name = "  North Harbour  "  # Simulates a place name with extra spacing.
response_count = int(raw_response_count)  # Converts the whole-number text into an integer for calculation.
file_label = raw_place_name.strip().lower().replace(" ", "_")  # Creates a simple standardised label while preserving the raw input.
print("Raw count:", raw_response_count)  # Displays the original text for traceability.
print("Converted count:", response_count)  # Displays the numerical working value.
print("Draft file label:", file_label)  # Displays the standardised text label.
```

## Exercise 2.7.1 — Integrated foundation script

**Case question:** A community programme needs a small, traceable summary created from supplied counts and a descriptive label.

**Code logic**

1. Preserve the supplied values.
2. Calculate complete and missing counts.
3. Calculate and format the completion rate.
4. Create a readable evidence statement.

```python
# Summarise one fictional monthly community-programme report.
programme_name = "Neighbourhood workshops"  # Stores the programme label used in the final statement.
total_records = 320  # Stores the total number of expected records.
missing_records = 24  # Stores the non-overlapping count of missing records.
complete_records = total_records - missing_records  # Calculates the records available for the technical summary.
completion_rate = (complete_records / total_records) * 100  # Calculates the complete share as a percentage.
evidence_statement = f"{programme_name}: {complete_records} of {total_records} records complete ({completion_rate:.1f}%)."  # Builds one traceable sentence from the stored values.
print(evidence_statement)  # Displays the final technical evidence statement.
```
