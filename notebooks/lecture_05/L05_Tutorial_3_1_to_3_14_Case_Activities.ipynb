## Model 3.14.4 — Create and inspect derived measures

The mean row rate gives every record equal weight, while the aggregate rate weights records through their case volumes. Three rows exceed 100% and one negative denominator makes a row result substantively invalid. Preserve these flags rather than hiding them.


````python
df["unresolved_cases"] = df["cases_received"] - df["cases_resolved"]  # Stores the calculated or supplied value for later use.
print(df["unresolved_cases"].nsmallest(5))  # Displays the result so it can be inspected.
display(df[df["unresolved_cases"] < 0][["record_id", "cases_received", "cases_resolved", "unresolved_cases"]])  # Displays the result so it can be inspected.
df["resolution_rate"] = df["cases_resolved"] / df["cases_received"] * 100  # Stores the calculated or supplied value for later use.
display(df[df["resolution_rate"] > 100][["record_id", "resolution_rate"]])  # Displays the result so it can be inspected.
print("Mean row rate:", round(df["resolution_rate"].mean(), 2))  # Displays the result so it can be inspected.
print("Aggregate rate:", round(df["cases_resolved"].sum() / df["cases_received"].sum() * 100, 2))  # Displays the result so it can be inspected.
````
