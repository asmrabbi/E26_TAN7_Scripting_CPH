### Exercise 2.15.1 — Allocate TAN7 project groups across two campuses

AAU teaches Techno-Anthropology in Aalborg and Copenhagen. Copenhagen has 45 students and Aalborg has 36 students, and groups must contain four or five students without mixing campuses. A smaller remainder must be reported rather than hidden. The report must verify every group size and total.

**Code plan**

1. Store campus counts.
2. Find a valid group count.
3. Distribute students across groups.
4. Validate sizes and totals.
5. Display the campus report and human-coordination note.

````python
campus_students = {"Copenhagen": 45, "Aalborg": 36}  # Stores the value needed by later steps.
def create_group_sizes(student_count):  # Defines a reusable function with explicit inputs.
    if student_count < 4:  # Starts a branch when this condition is true.
        return []  # Returns the result to the calling code.
    minimum_groups = (student_count + 4) // 5  # Stores the value needed by later steps.
    maximum_groups = student_count // 4  # Stores the value needed by later steps.
    if minimum_groups > maximum_groups:  # Starts a branch when this condition is true.
        return []  # Returns the result to the calling code.
    group_count = minimum_groups  # Stores the value needed by later steps.
    base_size = student_count // group_count  # Stores the value needed by later steps.
    extra_students = student_count % group_count  # Stores the value needed by later steps.
    group_sizes = []  # Stores the value needed by later steps.
    for group_index in range(group_count):  # Repeats the indented steps for each item.
        group_size = base_size + (1 if group_index < extra_students else 0)  # Stores the value needed by later steps.
        group_sizes.append(group_size)  # Runs this documented step in the solution.
    return group_sizes  # Returns the result to the calling code.
for campus, student_count in campus_students.items():  # Repeats the indented steps for each item.
    group_sizes = create_group_sizes(student_count)  # Stores the value needed by later steps.
    sizes_are_valid = all(4 <= group_size <= 5 for group_size in group_sizes)  # Runs this documented step in the solution.
    total_is_valid = sum(group_sizes) == student_count  # Runs this documented step in the solution.
    print(campus, "groups:", group_sizes)  # Displays a labelled result for inspection.
    print("Number of groups:", len(group_sizes))  # Displays a labelled result for inspection.
    print("Valid sizes and total:", sizes_are_valid and total_is_valid)  # Displays a labelled result for inspection.
print("Human coordination is still needed for accessibility, preferences and prior collaboration.")  # Displays a labelled result for inspection.
````

### Exercise 2.15.3 — Build a bounded course-keyword guessing activity

A lecturer wants a short activity for recognising Python keywords. The word is selected reproducibly, invalid entries do not consume attempts and the activity stops after six wrong letters. Correct letters are revealed after every accepted guess. The final message reports success or the revealed word.

**Code plan**

1. Import random and choose reproducibly.
2. Build the visible word.
3. Validate guesses.
4. Use a bounded loop.
5. Report the final state.

**Source/adaptation note:** Rodrigo Pinheiro’s hangman.py; rewritten for deterministic, bounded Colab execution.

````python
# Adapted from https://github.com/roedorpi/TAN7_Scripting_classnotes/blob/master/hangman.py
import random  # Imports the named standard-library tool for this step.
course_keywords = ["while", "return", "import"]  # Stores the value needed by later steps.
random_generator = random.Random(7)  # Stores the value needed by later steps.
secret_word = random_generator.choice(course_keywords)  # Stores the value needed by later steps.
simulated_guesses = ["r", "e", "1", "t", "u", "r", "n"]  # Stores the value needed by later steps.
guessed_letters = set()  # Stores the value needed by later steps.
wrong_attempts = 0  # Stores the value needed by later steps.
guess_index = 0  # Stores the value needed by later steps.
def visible_word(word, guesses):  # Defines a reusable function with explicit inputs.
    return " ".join(letter if letter in guesses else "_" for letter in word)  # Returns the result to the calling code.
while wrong_attempts < 6 and set(secret_word) - guessed_letters and guess_index < len(simulated_guesses):  # Repeats the indented steps while the condition remains true.
    guess = simulated_guesses[guess_index].strip().lower()  # Stores the value needed by later steps.
    guess_index += 1  # Stores the value needed by later steps.
    if len(guess) != 1 or not guess.isalpha() or guess in guessed_letters:  # Starts a branch when this condition is true.
        print("Ignored entry:", repr(guess))  # Displays a labelled result for inspection.
        continue  # Moves safely to the next loop iteration.
    guessed_letters.add(guess)  # Runs this documented step in the solution.
    if guess not in secret_word:  # Starts a branch when this condition is true.
        wrong_attempts += 1  # Stores the value needed by later steps.
    print(visible_word(secret_word, guessed_letters), "wrong attempts:", wrong_attempts)  # Displays a labelled result for inspection.
completed = not (set(secret_word) - guessed_letters)  # Stores the value needed by later steps.
print("Completed:", completed)  # Displays a labelled result for inspection.
print("Word:", secret_word)  # Displays a labelled result for inspection.
````

````python
# Adapted from https://github.com/roedorpi/TAN7_Scripting_classnotes/blob/master/session03_exercises.py
cookbook = {  # Starts a multi-line collection or expression.
    "lentil soup": [{"ingredient": "lentils", "quantity": 80, "unit": "g"}, {"ingredient": "water", "quantity": 250, "unit": "ml"}],  # Runs this documented step in the solution.
    "fruit salad": [{"ingredient": "apple", "quantity": 1, "unit": "piece"}, {"ingredient": "berries", "quantity": 60, "unit": "g"}],  # Runs this documented step in the solution.
}  # Closes the collection or expression started above.
def normalise_recipe_name(name):  # Defines a reusable function with explicit inputs.
    return name.strip().lower()  # Returns the result to the calling code.
def scale_recipe(name, portions):  # Defines a reusable function with explicit inputs.
    recipe_name = normalise_recipe_name(name)  # Stores the value needed by later steps.
    if recipe_name not in cookbook or portions <= 0:  # Starts a branch when this condition is true.
        return []  # Returns the result to the calling code.
    scaled_ingredients = []  # Stores the value needed by later steps.
    for ingredient in cookbook[recipe_name]:  # Repeats the indented steps for each item.
        scaled_ingredient = ingredient.copy()  # Stores the value needed by later steps.
        scaled_ingredient["quantity"] = ingredient["quantity"] * portions  # Stores the value needed by later steps.
        scaled_ingredients.append(scaled_ingredient)  # Runs this documented step in the solution.
    return scaled_ingredients  # Returns the result to the calling code.
selected_recipe = " LENTIL SOUP "  # Stores the value needed by later steps.
scaled_ingredients = scale_recipe(selected_recipe, 6)  # Stores the value needed by later steps.
if scaled_ingredients:  # Starts a branch when this condition is true.
    for ingredient in scaled_ingredients:  # Repeats the indented steps for each item.
        print(ingredient["ingredient"], ingredient["quantity"], ingredient["unit"])  # Displays a labelled result for inspection.
else:  # Starts the fallback branch.
    print("Choose one of:", sorted(cookbook))  # Displays a labelled result for inspection.
````

````python
# Adapted from https://github.com/roedorpi/TAN7_Scripting_classnotes/blob/master/session01_exercises.py
# Adapted from https://github.com/roedorpi/TAN7_Scripting_classnotes/blob/master/session05_functions_in_class.py
temperature_records = [  # Starts a multi-line collection or expression.
    {"place": "Harbour", "value": "50", "unit": " f "},  # Runs this documented step in the solution.
    {"place": "Campus", "value": "18", "unit": "C"},  # Runs this documented step in the solution.
    {"place": "Station", "value": "warm", "unit": "C"},  # Runs this documented step in the solution.
]  # Closes the collection or expression started above.
def to_celsius(value, unit):  # Defines a reusable function with explicit inputs.
    normalised_unit = unit.strip().upper()  # Stores the value needed by later steps.
    numerical_value = float(value)  # Stores the value needed by later steps.
    if normalised_unit == "C":  # Starts a branch when this condition is true.
        return numerical_value  # Returns the result to the calling code.
    if normalised_unit == "F":  # Starts a branch when this condition is true.
        return (numerical_value - 32) * 5 / 9  # Returns the result to the calling code.
    raise ValueError("Unknown temperature unit")  # Stops this function with a specific, informative error.
def classify_celsius(value):  # Defines a reusable function with explicit inputs.
    if value <= 0:  # Starts a branch when this condition is true.
        return "Freezing"  # Returns the result to the calling code.
    if value < 15:  # Starts a branch when this condition is true.
        return "Cool"  # Returns the result to the calling code.
    if value < 25:  # Starts a branch when this condition is true.
        return "Moderate"  # Returns the result to the calling code.
    return "Hot"  # Returns the result to the calling code.
converted_records = []  # Stores the value needed by later steps.
review_records = []  # Stores the value needed by later steps.
for record in temperature_records:  # Repeats the indented steps for each item.
    try:  # Starts a protected conversion or calculation block.
        celsius_value = to_celsius(record["value"], record["unit"])  # Stores the value needed by later steps.
    except ValueError as error:  # Handles the stated expected error.
        review_record = record.copy()  # Stores the value needed by later steps.
        review_record["reason"] = str(error)  # Stores the value needed by later steps.
        review_records.append(review_record)  # Runs this documented step in the solution.
        continue  # Moves safely to the next loop iteration.
    converted_record = record.copy()  # Stores the value needed by later steps.
    converted_record["celsius"] = round(celsius_value, 1)  # Stores the value needed by later steps.
    converted_record["classification"] = classify_celsius(celsius_value)  # Stores the value needed by later steps.
    converted_records.append(converted_record)  # Runs this documented step in the solution.
print("Converted:", converted_records)  # Displays a labelled result for inspection.
print("Review:", review_records)  # Displays a labelled result for inspection.
````
