### Exercise 2.8.1 — Classify consultation engagement

Classify engagement as Very high from 300, High from 200, Medium from 100 and Low otherwise. Test the values 99, 100, 199, 200, 299 and 300.

**Code plan**

1. Store the boundary tests.
2. Write a classification function.
3. Loop through the tests.
4. Display every input and result.

```python
boundary_values = [99, 100, 199, 200, 299, 300]  # Stores the value needed by later steps.
def classify_engagement(value):  # Defines a reusable function with explicit inputs.
    if value >= 300:  # Starts a branch when this condition is true.
        return "Very high"  # Returns the result to the calling code.
    if value >= 200:  # Starts a branch when this condition is true.
        return "High"  # Returns the result to the calling code.
    if value >= 100:  # Starts a branch when this condition is true.
        return "Medium"  # Returns the result to the calling code.
    return "Low"  # Returns the result to the calling code.
for boundary_value in boundary_values:  # Repeats the indented steps for each item.
    print(boundary_value, classify_engagement(boundary_value))  # Displays a labelled result for inspection.
```

### Exercise 2.10.1 — Validate numerical-looking input

Process the simulated inputs 25, text, -1 and 101 as percentages. Conversion errors and out-of-range values must have different messages.

**Code plan**

1. Loop through the raw text values.
2. Catch ValueError narrowly.
3. Validate the zero-to-one-hundred range.
4. Display each decision.

```python
raw_percentages = ["25", "text", "-1", "101"]  # Stores the value needed by later steps.
for raw_percentage in raw_percentages:  # Repeats the indented steps for each item.
    try:  # Starts a protected conversion or calculation block.
        percentage = int(raw_percentage)  # Stores the value needed by later steps.
    except ValueError:  # Handles the stated expected error.
        print(raw_percentage, "Conversion error")  # Displays a labelled result for inspection.
        continue  # Moves safely to the next loop iteration.
    if 0 <= percentage <= 100:  # Starts a branch when this condition is true.
        print(raw_percentage, "Valid percentage")  # Displays a labelled result for inspection.
    else:  # Starts the fallback branch.
        print(raw_percentage, "Range error")  # Displays a labelled result for inspection.
```

### Exercise 2.11.1 — Count categories in structured records

Count the categories in a list of observation dictionaries and identify unique campuses. Preserve the original records.

**Code plan**

1. Store a list of dictionaries.
2. Prepare a counter dictionary and campus set.
3. Loop through the records.
4. Display the counts and unique campuses.

```python
observations = [  # Starts a multi-line collection or expression.
    {"category": "Bus", "campus": "Aalborg"},  # Runs this documented step in the solution.
    {"category": "Cycle", "campus": "Copenhagen"},  # Runs this documented step in the solution.
    {"category": "Bus", "campus": "Copenhagen"},  # Runs this documented step in the solution.
]  # Closes the collection or expression started above.
category_counts = {}  # Stores the value needed by later steps.
campuses = set()  # Stores the value needed by later steps.
for observation in observations:  # Repeats the indented steps for each item.
    category = observation["category"]  # Stores the value needed by later steps.
    category_counts[category] = category_counts.get(category, 0) + 1  # Stores the value needed by later steps.
    campuses.add(observation["campus"])  # Runs this documented step in the solution.
print("Category counts:", category_counts)  # Displays a labelled result for inspection.
print("Campuses:", sorted(campuses))  # Displays a labelled result for inspection.
print("Raw records preserved:", observations)  # Displays a labelled result for inspection.
```

### Exercise 2.12.1 — Repeat until a valid menu choice

Use simulated inputs to model a prompt that repeats until the user chooses A, B or C. The loop must stop visibly and remain bounded in the notebook.

**Code plan**

1. Store simulated responses.
2. Track the current position.
3. Use while with an explicit bound.
4. Stop when a valid choice appears.

```python
simulated_choices = ["x", "", "b"]  # Stores the value needed by later steps.
choice_index = 0  # Stores the value needed by later steps.
accepted_choice = None  # Stores the value needed by later steps.
while choice_index < len(simulated_choices) and accepted_choice is None:  # Repeats the indented steps while the condition remains true.
    current_choice = simulated_choices[choice_index].strip().upper()  # Stores the value needed by later steps.
    choice_index += 1  # Stores the value needed by later steps.
    if current_choice in {"A", "B", "C"}:  # Starts a branch when this condition is true.
        accepted_choice = current_choice  # Stores the value needed by later steps.
    else:  # Starts the fallback branch.
        print("Rejected choice:", repr(current_choice))  # Displays a labelled result for inspection.
print("Accepted choice:", accepted_choice)  # Displays a labelled result for inspection.
```

### Exercise 2.13.1 — Reuse a validation function

Write a function that returns a completion percentage or None when the total is invalid. Apply it to several records and keep printing outside the function.

**Code plan**

1. Define the function with two parameters.
2. Validate the whole and part.
3. Return a reusable value.
4. Loop through tests and display results.

```python
completion_tests = [(72, 100), (0, 0), (105, 100)]  # Stores the value needed by later steps.
def completion_percentage(completed, invited):  # Defines a reusable function with explicit inputs.
    if invited <= 0 or completed < 0 or completed > invited:  # Starts a branch when this condition is true.
        return None  # Returns the result to the calling code.
    return completed / invited * 100  # Returns the result to the calling code.
for completed, invited in completion_tests:  # Repeats the indented steps for each item.
    result = completion_percentage(completed, invited)  # Stores the value needed by later steps.
    print(completed, invited, result)  # Displays a labelled result for inspection.
```

### Exercise 2.14.1 — Import, calculate and debug

**Code plan**

3. Test ordinary and empty lists.
