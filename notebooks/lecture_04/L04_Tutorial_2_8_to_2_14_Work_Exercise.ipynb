### Worked Exercise 2.8.35 — Classify an online fieldnote

**Story:** A Digital Anthropology group observes a public livestream about local cultural life.  
One student records how many minutes the group observed.  
The group calls 45 minutes or more an extended fieldnote, 20 to 44 minutes a standard fieldnote, and anything shorter a brief fieldnote.  
These labels organise the class notes; they do not measure the quality of the ethnography.

**Question:** Ask the student for the number of observation minutes and print the matching fieldnote label.

**Steps to solve it:**

1. Use `input()` to ask for the number of minutes.
2. Convert the answer to an integer.
3. Check the highest threshold first with `if`.
4. Check the second threshold with `elif`.
5. Use `else` for every smaller value and print the selected label.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested value.")  # Explains that the cell waits for keyboard input.
observation_minutes = int(input("Minutes observed: "))  # Collects and converts the observation length.

if observation_minutes >= 45:  # Checks the highest threshold first.
    print("Extended fieldnote")  # Displays the label for 45 minutes or more.
elif observation_minutes >= 20:  # Checks the second threshold after the first was false.
    print("Standard fieldnote")  # Displays the label for 20 to 44 minutes.
else:  # Handles every smaller value.
    print("Brief fieldnote")  # Displays the label for fewer than 20 minutes.
```

**Representative check:** `32` → `Standard fieldnote`

Test the exact boundaries as well as one value immediately below and above each boundary.


### Worked Exercise 2.8.36 — Review a stakeholder map

**Story:** A group in Framing Techno-Anthropological Transformation is preparing a fictional neighbourhood heat-plan case.  
They count the different stakeholder groups represented on their map.  
Eight or more groups is labelled broad, four to seven is developing, and fewer than four needs expansion.  
The labels help the group plan its next discussion; they do not prove that every voice is represented.

**Question:** Ask for the number of stakeholder groups and print the matching map label.

**Steps to solve it:**

1. Collect the group count with `input()` and convert it to an integer.
2. Test `8` or more first.
3. Use `elif` to test `4` or more.
4. Use `else` for a count below `4`.
5. Print one clear message for the group.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested value.")  # Explains that the cell waits for keyboard input.
stakeholder_groups = int(input("Number of stakeholder groups: "))  # Collects and converts the group count.

if stakeholder_groups >= 8:  # Checks the highest threshold first.
    print("Broad stakeholder map")  # Displays the label for eight groups or more.
elif stakeholder_groups >= 4:  # Checks the second threshold after the first was false.
    print("Developing stakeholder map")  # Displays the label for four to seven groups.
else:  # Handles every smaller count.
    print("Expand the stakeholder map")  # Displays the action for fewer than four groups.
```

### Worked Exercise 2.8.37 — Classify a wayfinding test

**Story:** A TAN7 group tests a fictional wayfinding kiosk before discussing the design.  
A participant tries to find the accessibility information, and the group records the time in seconds.  
Thirty seconds or less is labelled quick, 31 to 60 seconds is workable, and more than 60 seconds suggests revision.  
The time is one observation and does not by itself prove that the kiosk is usable.

**Question:** Ask for the completion time and print the matching test label.

**Steps to solve it:**

1. Ask for the number of seconds and convert the answer to an integer.
2. Check the lowest upper boundary first: `30` seconds or less.
3. Use `elif` for `60` seconds or less.
4. Use `else` for every longer time.
5. Print the result in plain language.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested value.")  # Explains that the cell waits for keyboard input.
completion_seconds = int(input("Seconds needed to find the information: "))  # Collects and converts the completion time.

if completion_seconds <= 30:  # Checks the lowest upper boundary first.
    print("Quick completion")  # Displays the label for 30 seconds or less.
elif completion_seconds <= 60:  # Checks the next upper boundary.
    print("Workable completion time")  # Displays the label for 31 to 60 seconds.
else:  # Handles every longer time.
    print("The wayfinding design needs revision")  # Displays the revision message.
```

**Representative check:** `52` → `Workable completion time`

Test the exact boundaries as well as one value immediately below and above each boundary.


### Worked Exercise 2.8.38 — Check a seven-day media diary

**Story:** A Digital Anthropology exercise asks a participant to make one media-diary entry on each of seven days.  
The student enters how many daily entries were completed.  
Seven entries is complete, four to six is usable but incomplete, and fewer than four is too limited for this classroom task.  
This teaching rule organises follow-up work; it is not a universal research standard.

**Question:** Ask for the number of completed days and print the diary status.

**Steps to solve it:**

1. Use `input()` and convert the number of completed days to an integer.
2. Use `if` with `==` to identify exactly seven entries.
3. Use `elif` with `>=` for four to six entries.
4. Use `else` for fewer than four entries.
5. Print one diary status.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested value.")  # Explains that the cell waits for keyboard input.
completed_days = int(input("Completed diary days: "))  # Collects and converts the completed-day count.

if completed_days == 7:  # Checks for the complete seven-day diary.
    print("Complete media diary")  # Displays the complete status.
elif completed_days >= 4:  # Checks for four to six completed days.
    print("Usable but incomplete media diary")  # Displays the middle status.
else:  # Handles fewer than four completed days.
    print("Media diary is too limited for this task")  # Displays the limited-data status.
```

**Representative check:** `5` → `Usable but incomplete media diary`

Test the exact boundaries as well as one value immediately below and above each boundary.


### Worked Exercise 2.8.39 — Interpret an AI explanation rating

**Story:** A class tests a fictional explanation shown after an automated application-sorting decision.  
One participant rates the explanation from `1` to `5`, where `5` is the highest rating.  
A rating of `5` is labelled very clear, `3` or `4` is partly clear, and a lower rating is unclear.  
One rating describes one response; it cannot establish that the system is fair.

**Question:** Ask for the clarity rating and print the matching interpretation.

**Steps to solve it:**

1. Ask for the rating and convert it to an integer.
2. Use `if` with `==` for the highest rating.
3. Use `elif` with `>=` for ratings `3` and `4`.
4. Use `else` for lower ratings.
5. Print the interpretation without making a fairness claim.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested value.")  # Explains that the cell waits for keyboard input.
clarity_rating = int(input("Explanation clarity rating (1-5): "))  # Collects and converts the rating.

if clarity_rating == 5:  # Checks for the highest rating.
    print("Very clear explanation")  # Displays the highest clarity label.
elif clarity_rating >= 3:  # Checks for ratings three and four.
    print("Partly clear explanation")  # Displays the middle clarity label.
else:  # Handles ratings below three.
    print("Unclear explanation")  # Displays the lowest clarity label.
```

**Representative check:** `3` → `Partly clear explanation`

### Worked Exercise 2.8.40 — Classify a moderation response

**Story:** A student studies a fictional online community with a published safety-report process.  
The student records how many hours passed before a moderator acknowledged one report.  
Six hours or less is labelled quick, more than six but no more than 24 hours is same-day, and a longer wait is delayed.  
Response speed is useful to record, but it does not show whether the moderation decision was fair.

**Question:** Ask for the response time in hours and print the matching response label.

**Steps to solve it:**

1. Ask for the number of hours and convert the answer to a float.
2. Test `6` hours or less first.
3. Use `elif` to test `24` hours or less.
4. Use `else` for a longer response time.
5. Print the label that matches the entered value.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested value.")  # Explains that the cell waits for keyboard input.
response_hours = float(input("Hours before acknowledgement: "))  # Collects and converts the response time.

if response_hours <= 6:  # Checks the quickest response range.
    print("Quick acknowledgement")  # Displays the quickest label.
elif response_hours <= 24:  # Checks the same-day response range.
    print("Same-day acknowledgement")  # Displays the middle label.
else:  # Handles every longer response time.
    print("Delayed acknowledgement")  # Displays the delayed label.
```

**Representative check:** `30` → `Delayed acknowledgement`

Test the exact boundaries as well as one value immediately below and above each boundary.
