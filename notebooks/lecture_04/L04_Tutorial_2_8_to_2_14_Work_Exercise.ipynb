### Worked Exercise 2.8.35 — Classify an online fieldnote

**Story:** A Digital Anthropology group observes a public livestream about local cultural life.  
One student records how many minutes the group observed.  
The group calls 45 minutes or more an extended fieldnote, 20 to 44 minutes a standard fieldnote, and anything shorter a brief fieldnote.  
These labels organise the class notes; they do not measure the quality of the ethnography.

**Question:** Ask the student for the number of observation minutes and print the matching fieldnote label.

**Steps to solve it:**

1. Use `input()` to ask for the number of minutes.
2. Convert the answer to an integer.
3. Check the highest threshold first with `if`.
4. Check the second threshold with `elif`.
5. Use `else` for every smaller value and print the selected label.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested value.")  # Explains that the cell waits for keyboard input.
observation_minutes = int(input("Minutes observed: "))  # Collects and converts the observation length.

if observation_minutes >= 45:  # Checks the highest threshold first.
    print("Extended fieldnote")  # Displays the label for 45 minutes or more.
elif observation_minutes >= 20:  # Checks the second threshold after the first was false.
    print("Standard fieldnote")  # Displays the label for 20 to 44 minutes.
else:  # Handles every smaller value.
    print("Brief fieldnote")  # Displays the label for fewer than 20 minutes.
```

**Representative check:** `32` → `Standard fieldnote`

### Worked Exercise 2.9.15 — Prepare a critical counter-map

**Story:** A Digital Anthropology group is preparing a small counter-map about places affected by a fictional redevelopment plan.<br>
The students enter how many local places appear on the map and whether they added a short note explaining the local context.<br>
The map is ready for discussion only when it contains at least four places and includes local context.<br>
If four places are present but the context is missing, the program should name that specific next step.

**Question:** Ask for the number of mapped places and whether local context was added, then print the correct preparation message.

**Steps to solve it:**

1. Collect and convert the place count.
2. Convert the yes/no context answer into a Boolean comparison.
3. Use `and` for the two readiness requirements.
4. Use `not` to identify missing context in `elif`.
5. Use `else` when more places are needed.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested values.")  # Explains that the cell waits for keyboard input.
mapped_places = int(input("Number of local places on the map: "))  # Collects and converts the number of places.
local_context_added = input("Was local context added? yes/no: ").strip().lower() == "yes"  # Converts the context answer to a Boolean.

if mapped_places >= 4 and local_context_added:  # Requires enough places and local context.
    print("Counter-map is ready for discussion")  # Displays the ready message.
elif mapped_places >= 4 and not local_context_added:  # Finds a map with enough places but missing context.
    print("Add local context before the discussion")  # Displays the specific next step.
else:  # Handles a map with fewer than four places.
    print("Map more local places first")  # Displays the remaining preparation task.
```

**Representative check:** `5`, `no` → `Add local context before the discussion`

### Worked Exercise 2.9.16 — Decide whether an archival item can be used

**Story:** A student finds a fictional digitised item while studying how the digital can work as an archive.<br>
The item may be available through public access or through university permission.<br>
A separate rights note can restrict classroom reuse even when one access route exists.<br>
The program must distinguish permitted use, a rights review, and missing access.

**Question:** Ask three yes/no questions and decide what the student should do with the archival item.

**Steps to solve it:**

1. Convert each yes/no answer into a Boolean.
2. Group the access routes with parentheses and `or`.
3. Add `and not rights_restricted` to apply the reuse restriction.
4. Use `elif` for the specific rights-review message.
5. Use `else` when access is missing.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested values.")  # Explains that the cell waits for keyboard input.
public_access = input("Is the item publicly accessible? yes/no: ").strip().lower() == "yes"  # Converts the public-access answer to a Boolean.
university_permission = input("Is university permission available? yes/no: ").strip().lower() == "yes"  # Converts the permission answer to a Boolean.
rights_restricted = input("Does the item have a reuse restriction? yes/no: ").strip().lower() == "yes"  # Converts the restriction answer to a Boolean.

if (public_access or university_permission) and not rights_restricted:  # Allows either access route when reuse is not restricted.
    print("The item may be used for the classroom task")  # Displays the permitted-use message.
elif rights_restricted:  # Gives the restriction priority over the access routes.
    print("Pause and review the rights note")  # Displays the rights-review message.
else:  # Handles the absence of both access routes.
    print("Access permission is needed")  # Displays the access message.
```

**Representative check:** `no`, `yes`, `no` → `The item may be used for the classroom task`

### Worked Exercise 2.9.17 — Respond to a platform outage

**Story:** A Digital Anthropology group treats a fictional campus platform as infrastructure and records a service interruption.<br>
The students enter whether an outage is confirmed, how many services are affected, and whether a backup channel is available.<br>
The impact question matters only after the outage has been confirmed, so the decision should be nested.<br>
Three affected services and no backup require the strongest response.

**Question:** Build a nested decision that prints the appropriate outage response.

**Steps to solve it:**

1. Keep the first yes/no answer as cleaned text so `yes`, `no` and another answer can take different branches.
2. In the outer `if`, continue only when the outage answer is `yes`.
3. Ask for the affected-service count and backup channel only inside that confirmed-outage branch.
4. Inside that branch, use `and`, `or` and `not` to distinguish urgent, priority and routine responses.
5. Use the outer `elif` and `else` for `no` and an unclear answer.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested values.")  # Explains that the cell waits for keyboard input.
outage_answer = input("Is the outage confirmed? yes/no: ").strip().lower()  # Cleans the outage answer while preserving yes, no or another value.

if outage_answer == "yes":  # Opens the impact questions only after an outage is confirmed.
    affected_services = int(input("Number of affected services: "))  # Collects and converts the service count inside the confirmed-outage branch.
    backup_available = input("Is a backup channel available? yes/no: ").strip().lower() == "yes"  # Converts the backup answer inside the confirmed-outage branch.

    if affected_services >= 3 and not backup_available:  # Finds the strongest combination of broad impact and no backup.
        print("Urgent infrastructure response")  # Displays the urgent response.
    elif affected_services >= 3 or not backup_available:  # Finds either broad impact or a missing backup.
        print("Priority infrastructure review")  # Displays the priority response.
    else:  # Handles a smaller outage with a backup.
        print("Routine outage documentation")  # Displays the routine response.
elif outage_answer == "no":  # Handles a clear statement that no outage is confirmed.
    print("Record that no outage is confirmed")  # Displays the no-outage message.
else:  # Handles an answer other than yes or no.
    print("Enter yes or no for the outage question")  # Displays a focused correction.
```

**Representative check:** `yes`, `4`, `no` → `Urgent infrastructure response`

Run the alternative tests in the Checks section so every branch is exercised.


### Worked Exercise 2.9.18 — Apply a responsible-innovation stop rule

**Story:** A class discusses a fictional AI pilot using themes from Responsible Innovation and AI ethics.<br>
The students record whether a possible harm has been reported, whether human review is complete, and whether an appeal route exists.<br>
A harm signal or missing human review must pause the pilot before other conditions are considered.<br>
Only a reviewed pilot with an appeal route receives the continuation message.

**Question:** Ask the three questions and print whether to pause, continue carefully, or add an appeal route.

**Steps to solve it:**

1. Convert the three answers into Boolean values.
2. Put the stop rule first and join its alternatives with `or`.
3. Use `not` for missing human review.
4. Use `and` in `elif` for the continuation requirements.
5. Use `else` for a missing appeal route.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested values.")  # Explains that the cell waits for keyboard input.
harm_reported = input("Has a possible harm been reported? yes/no: ").strip().lower() == "yes"  # Converts the harm answer to a Boolean.
human_review_complete = input("Is human review complete? yes/no: ").strip().lower() == "yes"  # Converts the review answer to a Boolean.
appeal_route_available = input("Is an appeal route available? yes/no: ").strip().lower() == "yes"  # Converts the appeal answer to a Boolean.

if harm_reported or not human_review_complete:  # Gives either stop condition first priority.
    print("Pause the AI pilot for review")  # Displays the pause message.
elif human_review_complete and appeal_route_available:  # Requires both safeguards for limited continuation.
    print("The limited pilot may continue with monitoring")  # Displays the continuation message.
else:  # Handles a reviewed pilot without an appeal route.
    print("Add an appeal route before continuing")  # Displays the remaining safeguard.
```

**Representative check:** `no`, `yes`, `no` → `Add an appeal route before continuing`

Run the alternative tests in the Checks section so every branch is exercised.


### Worked Exercise 2.9.19 — Check a digital-participation plan

**Story:** A group designs a fictional public discussion inspired by the Digital Participation topic.<br>
People may join through an online session or a room-based session, and the information should be available in an accessible format.<br>
The first decision asks whether neither participation route exists by applying `not` to the grouped alternatives.<br>
If a route exists, the next decision checks whether the accessible material is ready.

**Question:** Ask about the two participation routes and the accessible material, then print the plan's next action.

**Steps to solve it:**

1. Convert all three answers into Boolean values.
2. Put the participation routes in parentheses with `or`.
3. Apply `not` to the grouped routes.
4. Use `elif` to check the accessible material after a route exists.
5. Use `else` when a route exists but the material is missing.

**Solution:** Run the next cell manually.


```python
print("Run this cell manually and enter the requested values.")  # Explains that the cell waits for keyboard input.
online_session = input("Is an online session available? yes/no: ").strip().lower() == "yes"  # Converts the online-session answer to a Boolean.
room_session = input("Is a room-based session available? yes/no: ").strip().lower() == "yes"  # Converts the room-session answer to a Boolean.
accessible_material = input("Is accessible information ready? yes/no: ").strip().lower() == "yes"  # Converts the accessibility answer to a Boolean.

if not (online_session or room_session):  # Checks whether both participation routes are absent.
    print("Create at least one participation route")  # Displays the missing-route action.
elif accessible_material:  # Checks accessibility only after a route exists.
    print("Participation plan is ready to review")  # Displays the ready message.
else:  # Handles an available route with missing accessible information.
    print("Add accessible information before inviting participants")  # Displays the accessibility action.
```

**Representative check:** `yes`, `no`, `no` → `Add accessible information before inviting participants`
